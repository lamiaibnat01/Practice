# **Environment Setup & Imports**

```python
import pandas as pd
from datetime import datetime

# Display settings
pd.set_option('display.float_format', '{:,.2f}'.format)

print(" Environment setup completed!")
print(f"Current Date & Time: {datetime.now().strftime('%Y-%m-%d %H:%M')}")
```

# **Input Section Python**

```python
print("=== Smart Salary & Bonus Calculator ===\n")

name = input("Enter your Name: ").strip()

age = int(input("Enter your Age: "))
monthly_salary = float(input("Enter your Monthly Salary ($): "))
experience = int(input("Enter your Experience (years): "))

print("\nInputs recorded successfully!")
```

# **Calculations**

```python

yearly_salary = monthly_salary * 12


bonus_percentage = 0.20 if experience >= 5 else 0.10
bonus = yearly_salary * bonus_percentage
bonus_message = "20% (Experienced)" if experience >= 5 else "10% (Standard)"


tax_percentage = 0.15 if yearly_salary > 50000 else 0.05
tax = yearly_salary * tax_percentage
tax_message = "15% (High Bracket)" if yearly_salary > 50000 else "5% (Standard)"


net_salary = yearly_salary + bonus - tax


senior_benefit = "Eligible" if age >= 40 else "Not Eligible"


earner_status = "High Earner" if yearly_salary > 80000 else "Regular Earner"

print(" All calculations completed!")
```

# **Results Display**

```python
print("="*60)
print("                 FINAL REPORT")
print("="*60)
print(f"Name                    : {name.title()}")
print(f"Age                     : {age} years")
print(f"Experience              : {experience} years")
print(f"Monthly Salary          : ${monthly_salary:,.2f}")
print(f"Yearly Salary           : ${yearly_salary:,.2f}")
print(f"Bonus ({bonus_message})     : ${bonus:,.2f}")
print(f"Tax ({tax_message})        : ${tax:,.2f}")
print(f"Net Yearly Income       : ${net_salary:,.2f}")
print(f"Senior Benefit          : {senior_benefit}")
print(f"Earner Status           : {earner_status}")
print("="*60)
```

# **Summary Table**

```python
data = {
    'Particulars': ['Yearly Salary', 'Bonus', 'Tax', 'Net Income'],
    'Amount ($)': [yearly_salary, bonus, tax, net_salary],
    'Percentage': ['100%', f'{bonus_percentage*100:.0f}%', f'{tax_percentage*100:.0f}%', '—']
}

summary_df = pd.DataFrame(data)
summary_df
```
