# **User Input**

```python
name = input("Enter Name: ")
age = int(input("Enter Age: "))
salary = float(input("Enter Monthly Salary: "))
experience = int(input("Enter Experience (Years): "))
```

# **Yearly Salary**


```python
yearly_salary = salary * 12

print("Yearly Salary:", yearly_salary)
```

# **Tax**

```python
if salary > 50000:
    tax = salary * 0.15
else:
    tax = salary * 0.05

print("Tax:", tax)
```

# **Bonus**

```python
if experience >= 5:
    bonus = salary * 0.20
else:
    bonus = salary * 0.10

print("Bonus:", bonus)
```

# **Senior Benefit**

```python
if age >= 40:
    senior_benefit = "Eligible"
else:
    senior_benefit = "Not Eligible"

print("Senior Benefit:", senior_benefit)
```

# **Status**

```python
status = "High Earner" if salary > 80000 else "Regular Earner"

print("Status:", status)
```

# **Final Output**

```python
print("\n===== FINAL REPORT =====")

print("Name:", name)
print("Age:", age)
print("Monthly Salary:", salary)
print("Experience:", experience)

print("Yearly Salary:", yearly_salary)
print("Tax:", tax)
print("Bonus:", bonus)

print("Senior Benefit:", senior_benefit)
print("Status:", status)
```
